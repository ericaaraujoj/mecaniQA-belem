# MecâniQA — Encontro 1: Importação e Inspeção Inicial

```python
import pandas as pd

# Importação do dataset
df = pd.read_csv("mecaniqa_dataset.csv")

print("Dataset importado com sucesso!")
df.head()
```

```python
# Remove possíveis espaços dos nomes das colunas
df.columns = df.columns.str.strip()

# Converte a coluna Data para datetime
df["Data"] = pd.to_datetime(
    df["Data"],
    format="%Y-%m-%d",
    errors="raise"
)

# Define Data como índice e organiza cronologicamente
df = df.set_index("Data")
df = df.sort_index()

print("Indexação temporal concluída!")
df.head()
```

```python
df.info()
```

```python
print("Quantidade de registros:", df.shape[0])
print("Quantidade de colunas:", df.shape[1])
print("Atributos principais:", [df.index.name] + df.columns.tolist())
print("Tipo do índice:", type(df.index))
```

```python
print("Quantidade de datas inválidas:", df.index.isna().sum())
print("Data inicial:", df.index.min())
print("Data final:", df.index.max())
```

```python
print("Valores ausentes por atributo:")
print(df.isnull().sum())
```
